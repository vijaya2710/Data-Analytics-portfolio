# **Supply Chain Shipments Dataset**

Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Loading Dataset

```python
df = pd.read_csv('/content/sample_data/supply_chain_shipments (1).csv')
df.head(5)
```

Required Preprocessing and KPI Calculations

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['month'] = df['order_date'].dt.month
df['on_time'] = df['actual_lead_days'] <= df['promised_lead_days']
df['in_full'] = (df['qty_delivered'] == df['qty_ordered']) & (df['is_damaged'] == 0)
df['otif'] = df['on_time'] & df['in_full']
```

Setting theme

```python
bg_color = '#fdfdfd'
grid_color = '#e0e0e0'
c_line = '#2b7fb3'
c_highlight = '#ffecd2'
c_scatter = ['#4f6367', '#7a9e9f', '#b8d8d8', '#fe5f55']
heatmap_camp = 'YlGnBu'
c_bar = ['#3b5a9d', '#49a078', '#dcc7be']
```

```python
plt.rcParams.update({
    'figure.facecolor' : bg_color,
    'axes.facecolor' : bg_color,
    'grid.color' : grid_color,
    'axes.edgecolor' : grid_color,
    'axes.grid' : True,
    'axes.axisbelow' : True
})
print("Setup & Data Preprocessing Sucessfully Completed")
```

2. Monthly OTIF% Trajectory (Line Chart)

```python
plt.figure(figsize = (9, 5))
monthly_otif = df.groupby('month')['otif'].mean() * 100

sns.lineplot(x = monthly_otif.index, y = monthly_otif.values, color = c_line, marker = 'o', linewidth = 2.5)
plt.axvspan(7, 9, color = c_highlight, alpha = 0.5, label = 'Monsoon Dips')

plt.title('Monthly OTIF % Trajectory', fontsize = 16, pad = 15)
plt.xlabel('Month')
plt.ylabel('OTIF %')
plt.xticks(range(1, 13))
plt.legend(loc = 'lower left')
plt.tight_layout()
plt.show()
```

3. Carrier Efficiency : Freight vs OTIF

```python
plt.figure(figsize = (9, 5))
carrier_stats = df.groupby('carrier_id').agg(
    avg_freight = ('freight_cost_inr', 'mean'),
    otif_pct = ('otif', lambda x: x.mean() * 100)
).reset_index()

sns.scatterplot(data = carrier_stats, x = 'avg_freight', y = 'otif_pct', hue = 'carrier_id', palette = c_scatter, s = 150)

plt.title('Carrier Efficiency : Freight vs OTIF', fontsize = 14, pad = 15)
plt.xlabel('Average Freight Cost (₹)')
plt.ylabel('OTIF (%)')

for i, row in carrier_stats.iterrows():
  plt.text(row['avg_freight'] + 15, row['otif_pct'], row['carrier_id'], fontsize = 10)

plt.legend().remove()
plt.tight_layout()
plt.show()
```

4. Regional Corridor OTIF

```python
plt.figure(figsize = (9,5))
heatmap_data = df.groupby(['origin_region', 'dest_region'])['otif'].mean().unstack()

sns.heatmap(heatmap_data, cmap = heatmap_camp, annot = True, fmt = '.1f', cbar_kws = {'label' : 'OTIF %'})

plt.title('Regional Corridor OTIF Heatmap', fontsize = 14, pad = 15)
plt.xlabel('Destination Origin')
plt.ylabel('Origin Region')
plt.grid(False)
plt.tight_layout()
plt.show()
```

5. Cost Structure Breakdown by Category

```python
cost_stats = df.groupby('category')[['freight_cost_inr', 'warehouse_handling_inr', 'return_penalty_inr']].mean().reset_index()

ax = cost_stats.set_index('category').plot(kind = 'bar', stacked = True, color = c_bar, figsize = (9,5))

plt.title('Cost Structure Breakdown By Category', fontsize = 14, pad = 15)
plt.xlabel('Category')
plt.ylabel('Cost Per Order (₹)')
plt.xticks(rotation = 0)
plt.legend(['Freight Cost', 'Warehouse Handling', 'Return Penalty'])
plt.grid(True)
plt.tight_layout()
plt.show()
```
